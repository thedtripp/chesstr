```python
import io
import chess.pgn
import math

with open("grand_prix_attack_gothamchess.pgn") as pgn:
    game = chess.pgn.read_game(pgn)

#display(game.board())
# print(type(game)) # <class 'chess.pgn.Game'>

def iterative_bfs(game, move):
    queue = [(game, 1)]
    while queue:
        # get node
        current, level = queue.pop(0)
        if current.next():
            display(current.board())
            # process node
            print(math.ceil(level/2), current.variations)
            print(math.ceil(level/2), current.next().san())
            print()

            # push children to stack
            for v in current.variations:
                queue.append((v, level + 1))
    
iterative_bfs(game, 1)
```
